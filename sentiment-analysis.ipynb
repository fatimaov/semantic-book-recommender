```python
# Load the tools and book descriptions needed for sentiment analysis.
import pandas as pd

books = pd.read_csv("data/books_with_categories.csv")
```

```python
# Create an emotion classifier and check it with a sample sentence.
from transformers import pipeline
classifier = pipeline("text-classification",
                      model="j-hartmann/emotion-english-distilroberta-base",
                      top_k = None,
                      device =0)
classifier("I love this!")
```

```python
# Display the first book description for analysis.
books["description"][0]
```

```python
# Classify the emotion expressed across the full description.
classifier(books["description"][0])
```

```python
# Compare emotion scores for each sentence in the description.
classifier(books["description"][0].split("."))
```

```python
# Split the description into sentences and classify each one.
sentences = books["description"][0].split(".")
predictions = classifier(sentences)
```
